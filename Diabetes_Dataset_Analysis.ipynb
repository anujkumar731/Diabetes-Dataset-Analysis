```python
# Load the Pima Indians Diabetes dataset

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/cs-helpdesk/datasets/main/diabetes.csv"

df = pd.read_csv(url)

# Display first five rows
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
# Check missing values
print("Missing values:")
print(df.isnull().sum())

# Check zero values in each column
print("\nZero values:")
print((df == 0).sum())
```

```python
# Check duplicate records
duplicates = df.duplicated().sum()

print("Total duplicate rows:", duplicates)

# Display duplicate rows if any
df[df.duplicated()]
```

```python
# Statistical summary
df.describe()
```

```python
# Check duplicate records
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Create a copy of original dataset
df_clean = df.copy()

# Columns where zero represents a missing measurement
zero_columns = [
    'Glucose',
    'BloodPressure',
    'SkinThickness',
    'Insulin',
    'BMI'
]

# Replace zero with NaN in selected columns
df_clean[zero_columns] = df_clean[zero_columns].replace(0, np.nan)

# Check missing values after replacement
print(df_clean.isnull().sum())
```

```python
# Calculate missing values
missing_count = df_clean.isnull().sum()

# Calculate missing percentage
missing_percentage = (
    df_clean.isnull().mean() * 100
).round(2)

# Create summary table
missing_summary = pd.DataFrame({
    'Missing Values': missing_count,
    'Missing Percentage': missing_percentage
})

display(missing_summary)
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x=missing_summary.index,
    y=missing_summary['Missing Percentage']
)

plt.title('Missing Values Percentage by Column')
plt.xlabel('Columns')
plt.ylabel('Missing Values (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Count diabetes outcomes
outcome_counts = df['Outcome'].value_counts()

print("Outcome Distribution:")
print(outcome_counts)

# Visualize outcome distribution
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='Outcome')

plt.title('Diabetes Outcome Distribution')
plt.xlabel('Outcome (0 = No Diabetes, 1 = Diabetes)')
plt.ylabel('Number of People')

plt.show()
```

```python
# Select important numerical features
features = ['Glucose', 'BMI', 'Age']

# Replace invalid zero measurements with NaN for visualization
plot_data = df[features].copy()
plot_data[['Glucose', 'BMI']] = (
    plot_data[['Glucose', 'BMI']].replace(0, np.nan)
)

# Create histograms
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for i, feature in enumerate(features):
    sns.histplot(
        plot_data[feature].dropna(),
        kde=True,
        ax=axes[i]
    )

    axes[i].set_title(f'Distribution of {feature}')
    axes[i].set_xlabel(feature)
    axes[i].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
# Calculate correlation between numerical columns
correlation = df_clean.corr(numeric_only=True)

# Display correlation values
display(correlation)
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Diabetes Dataset')
plt.tight_layout()
plt.show()
```

```python
# Compare features across diabetes outcomes

features = ['Glucose', 'BMI', 'Age']

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for i, feature in enumerate(features):

    sns.boxplot(
        data=df_clean,
        x='Outcome',
        y=feature,
        ax=axes[i]
    )

    axes[i].set_title(f'{feature} by Diabetes Outcome')
    axes[i].set_xlabel('Outcome (0 = No Diabetes, 1 = Diabetes)')
    axes[i].set_ylabel(feature)

plt.tight_layout()
plt.show()
```

```python
# Calculate average values for each outcome group

comparison = df_clean.groupby('Outcome')[
    ['Glucose', 'BMI', 'Age']
].mean().round(2)

print("Average Feature Values by Outcome:")
display(comparison)
```
